```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df_aquario = pd.read_csv('//content/aquario_publico_boxplot.csv')
display(df_aquario)
```

```python
plt.figure(figsize=(11,5))
plt.bar(df_aquario['especie'], df_aquario['comprimento_cm'])
plt.show()
```

```python
# agora vamos trazer o boxplot
plt.figure(figsize=(15,5))
df_aquario.boxplot(column=['comprimento_cm'], by='especie')
plt.show()
```
