```python
# https://docs.google.com/spreadsheets/d/13JRSvmPCVYWaNaMJedr83bYPcQHj57IPx0nWw-rlHo4/edit?usp=sharing

import pandas as pd
import requests
import time

endpoint_planilha = "https://docs.google.com/spreadsheets/d/13JRSvmPCVYWaNaMJedr83bYPcQHj57IPx0nWw-rlHo4/export?format=csv"
endpoint_api = "https://api.open-meteo.com/v1/forecast"
```

```python
print('-----Dados Coletados-----')
dados_planilha = pd.read_csv(endpoint_planilha)

display(dados_planilha)
```

```python
# iterar cada resposta do formulario

for indice, linha in dados_planilha.iterrows():
  nome_local = linha['Nome do local']
  Iat = linha['Latitude']
  Ion = linha['Longitude']

  parametros = {
      'latitude': Iat,
      'longitude' : Ion,
      'current' : 'temperature_2m'
  }
  resposta = requests.get(endpoint_api, params=parametros).json()


  display(resposta)
  print('\n\n')
```
