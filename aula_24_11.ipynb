```python
import pandas as pd
```

```python
df_beneficios = pd.read_csv('/content/beneficios_saude.csv')
df_clientes = pd.read_csv('/content/clientes.csv')
df_compras = pd.read_csv('/content/compras.csv')
```

```python
display(df_beneficios.head(1))
display(df_clientes.head(1))
display(df_compras.head(1))
```

```python
# vamos juntar as tabelas em uma só
# temos que fazer em duas etapas
# sempre as tabelas dão merge em pares
df_beneficios_clientes = pd.merge(df_clientes,
                                  df_beneficios,
                                  on='cpf')


df_beneficios_clientes_compras = pd.merge(df_beneficios_clientes,
                                          df_compras,
                                          on='cliente_id')



display(df_beneficios_clientes_compras)
```

```python
# removendo dados sem necessidade de exibição
df_dados_lgpd = df_beneficios_clientes_compras[['cliente_id',
                                                'idade', 'cep',
                                                'plano_saude',
                                                'condicao_cronica',
                                                'transacao_id',
                                                'valor_gasto',
                                                'categoria']]

display(df_dados_lgpd)


# aplicação de mascara
df_dados_lgpd['regiao'] = df_dados_lgpd['cep'].str[:3] + "xx-xxx"
display(df_dados_lgpd)

# regra de faixa
bins = [0,26,36,51,120]
faixas = ["18-25", "26-35", "36,50", "50+"]
df_dados_lgpd['faixa_etaria'] = pd.cut(df_dados_lgpd['idade'], bins=bins,
                                       labels=faixas)
display(df_dados_lgpd)
```
